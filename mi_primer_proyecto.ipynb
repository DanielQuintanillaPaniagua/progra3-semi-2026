```python
sueldo_bruto = lambda hrs, valor_hora: hrs * valor_hora * 30
calcular_isss = lambda sb: 30.0 if sb >= 1000.0 else sb * 0.03
calcular_afp = lambda sb: sb * 0.0725
calcular_isr = lambda r: (
    0.0 if r <= 472.00 else
    (r - 472.00) * 0.10 + 17.67 if r <= 895.24 else
    (r - 895.24) * 0.20 + 60.00 if r <= 2038.10 else
    (r - 2038.10) * 0.30 + 288.57)
```

```python
hrs_diarias = float(input("Ingrese horas trabajadas al día: "))
valor_hora = float(input("Ingrese valor por hora ($): "))

s_bruto = sueldo_bruto(hrs_diarias, valor_hora)
isss = calcular_isss(s_bruto)
afp = calcular_afp(s_bruto)

renta_gravable = s_bruto - isss - afp
isr = calcular_isr(renta_gravable)

total_deducciones = isss + afp + isr
sueldo_neto = s_bruto - total_deducciones

print(f"Sueldo Bruto: ${s_bruto:.2f}")
print(f"ISSS: ${isss:.2f}")
print(f"AFP: ${afp:.2f}")
print(f"ISR: ${isr:.2f}")
print(f"Total Deducciones: ${total_deducciones:.2f}")
print(f"Sueldo Neto a pagar: ${sueldo_neto:.2f}")
```

```python
semana = ["domingo", "lunes", "Martes", "Miercoles", "jueves", "Viernes", "sabado"]
print(semana,"Primer dia de la semana: ", semana[0])

primeraLetraDia = map(lambda dia: dia[0], semana)
print(list(primeraLetraDia))

primeraLetraMayusDia = map(lambda dia: dia[0].upper(), semana)
print(list(primeraLetraMayusDia))
```

```python
serie = [2,5,6,8,9,10,12]

pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares: ", list(pares))
print("Impares: ", list(impares))
```

```python
notas = input("Notas: ")
notas = notas.split(",")
aprobados = filter(lambda nota: float(nota)>=6, notas)
reprobados = filter(lambda nota: float(nota)<6, notas)
print("Notas: ", notas)
print("Aprobados: ", list(aprobados))
print("Reprobados: ", list(reprobados))
```

```python
from functools import reduce

numeros = [1,2,10,4,9,4]
pares = filter(lambda num: num%2==0, numeros)

print("Numeros:" , numeros)
#print("Pares: ", list(pares))

sumaPares = reduce(lambda acum, num: num+acum, pares, 0)
print("Suma de pares: ", sumaPares)
```
