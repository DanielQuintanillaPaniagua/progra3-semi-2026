```python

TABLA_IMPUESTOS = [
    (0.01,       500,        1.5,  0),
    (500.01,     1000,       1.5,  3),
    (1000.01,    2000,       3,    3),
    (2000.01,    3000,       6,    3),
    (3000.01,    6000,       9,    2),
    (6000.01,    18000,      15,   2),
    (18000.01,   30000,      39,   2),
    (30000.01,   60000,      63,   1),
    (60000.01,   100000,     93,   0.8),
    (100000.01,  200000,     125,  0.7),
    (200000.01,  300000,     195,  0.6),
    (300000.01,  400000,     255,  0.45),
    (400000.01,  500000,     300,  0.4),
    (500000.01,  1000000,    340,  0.30),
    (1000000.01, 99999999,   490,  0.18),
]


def calcular_impuesto(monto, tabla):
    for desde, hasta, precio, adicional in tabla:
        if desde <= monto <= hasta:
            pago = (monto - desde) / 1000 * adicional + precio
            return round(pago, 2)
    return None


def main():
    monto = float(input("Ingrese el monto de la actividad económica: $"))
    pago = calcular_impuesto(monto, TABLA_IMPUESTOS)

    if pago is not None:
        print(f"El valor a pagar es: ${pago}")
    else:
        print("El monto ingresado está fuera de los rangos de la tabla.")


if __name__ == "__main__":
    main()

```

```python
import math
from functools import reduce

class SistemaEstadistico:
    def __init__(self, datos_estudiantes):
        self.dataset = datos_estudiantes

    def obtener_columna(self, llave_campo):
        return [float(estudiante[llave_campo]) for estudiante in self.dataset.values()]

    def media_aritmetica(self, serie):
        if not serie: return 0
        suma = reduce(lambda acum, num: acum + num, serie, 0)
        return suma / len(serie)

    def mediana(self, serie):
        if not serie: return 0
        datos_ordenados = sorted(serie)
        n = len(datos_ordenados)
        mitad = n // 2

        if n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
        else:
            return datos_ordenados[mitad]

    def frecuencia(self, serie):
        fre = {}
        for num in serie:
            fre[num] = fre.get(num, 0) + 1
        return fre

    def moda(self, serie):
        if not serie: return "No hay datos"
        frecuencia_datos = self.frecuencia(serie)
        max_frecuencia = max(frecuencia_datos.values())

        if max_frecuencia == 1:
            return "No hay moda (Amodal)"

        modas = [num for num, freq in frecuencia_datos.items() if freq == max_frecuencia]
        return modas

    def varianza(self, serie):
        if len(serie) < 2: return 0
        media = self.media_aritmetica(serie)
        suma_cuadrados = sum((x - media) ** 2 for x in serie)
        return suma_cuadrados / len(serie)

    def desviacion_estandar(self, serie):
        return math.sqrt(self.varianza(serie))

    def desviacion_tipica(self, serie):
        return self.desviacion_estandar(serie)

    def rango(self, serie):
        if not serie: return 0
        return max(serie) - min(serie)

    def calcular_todo(self, nombre_campo):
        serie = self.obtener_columna(nombre_campo)
        print(f"\n=== ANÁLISIS ESTADÍSTICO: {nombre_campo.upper()} ===")
        print(f"Media Aritmética: {self.media_aritmetica(serie):.2f}")
        print(f"Mediana:          {self.mediana(serie):.2f}")
        print(f"Moda:             {self.moda(serie)}")
        print(f"Varianza:         {self.varianza(serie):.2f}")
        print(f"Desviación Est.:  {self.desviacion_estandar(serie):.2f}")
        print(f"Desviación Típ.:  {self.desviacion_tipica(serie):.2f}")
        print(f"Rango:            {self.rango(serie):.2f}")
        print(f"Frecuencias:      {self.frecuencia(serie)}")


dataset_ugb = {
    "estudiante_1":  {"nombre": "Samuel Alexander Portillo Cerón",   "materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 10.0, "edad": 20, "traslado_ugb": 80, "uso_telefono": 6.0},
    "estudiante_2":  {"nombre": "Carlos David Turcios Rivas",         "materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 9.5,  "edad": 20, "traslado_ugb": 25, "uso_telefono": 5.0},
    "estudiante_3":  {"nombre": "Gilda Mileidy González Castro",      "materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 9.38, "edad": 19, "traslado_ugb": 30, "uso_telefono": 5.92},
    "estudiante_4":  {"nombre": "Saúl Alexander Jimenez Garcia",      "materia": "Programación Computacional II, Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 8.4, "edad": 21, "traslado_ugb": 60, "uso_telefono": 5.0},
    "estudiante_5":  {"nombre": "Edgar Ernesto Ramirez Barrera",      "materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 10.0, "edad": 21, "traslado_ugb": 120, "uso_telefono": 7.0},
    "estudiante_6":  {"nombre": "Yeimy Julissa Parada Chávez",        "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.5, "edad": 20, "traslado_ugb": 45, "uso_telefono": 3.0},
    "estudiante_7":  {"nombre": "Miguel Alejandro Gómez Saravia",     "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.5, "edad": 19, "traslado_ugb": 15, "uso_telefono": 4.0},
    "estudiante_8":  {"nombre": "Josué Stiven Funes Benavides",       "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.5, "edad": 19, "traslado_ugb": 30, "uso_telefono": 5.0},
    "estudiante_9":  {"nombre": "Jefferson Steven Castillo Leiva",    "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.5, "edad": 19, "traslado_ugb": 15, "uso_telefono": 2.3},
    "estudiante_10": {"nombre": "Ramael Ernesto Cordova Colindres",   "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.5, "edad": 22, "traslado_ugb": 7, "uso_telefono": 13.52},
    "estudiante_11": {"nombre": "Carlos Daniel Parada Avalos",        "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 8.5, "edad": 25, "traslado_ugb": 10, "uso_telefono": 5.52},
    "estudiante_12": {"nombre": "Jennifer Maricela Renderos Guevara", "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 9.6, "edad": 21, "traslado_ugb": 60, "uso_telefono": 3.0},
    "estudiante_13": {"nombre": "Madison Odalis Valle Juárez",        "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 10.0, "edad": 20, "traslado_ugb": 30, "uso_telefono": 9.0},
    "estudiante_14": {"nombre": "Jose Luis Garcia Mejia",             "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 8.6, "edad": 20, "traslado_ugb": 90, "uso_telefono": 8.0},
    "estudiante_15": {"nombre": "Jaime Cha",                          "materia": "Programación Computacional III, Estadística Computacional, Diseño de Circuitos Digitales, Base de Datos II, Inglés Básico", "nota": 9.2, "edad": 19, "traslado_ugb": 7, "uso_telefono": 6.0},
    "estudiante_16": {"nombre": "Diego Misael Villafranco Pineda",    "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 8.0, "edad": 20, "traslado_ugb": 30, "uso_telefono": 8.0},
    "estudiante_17": {"nombre": "Flor Zulema Campos Orellana",        "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 8.0, "edad": 19, "traslado_ugb": 6, "uso_telefono": 5.38},
    "estudiante_18": {"nombre": "Kevin Campos",                       "materia": "Base de Datos II, Inglés Básico, Programación Computacional III, Diseño de Circuitos Digitales, Estadística Computacional", "nota": 8.6, "edad": 19, "traslado_ugb": 30, "uso_telefono": 6.0},
    "estudiante_19": {"nombre": "Wilfredo Josué Ortez Gaitán",        "materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 9.54, "edad": 20, "traslado_ugb": 30, "uso_telefono": 5.0},
    "estudiante_20": {"nombre": "Daniel Enrique Quintanilla Paniagua","materia": "Estadística, Diseño de Circuitos Digitales, Inglés Básico, Base de Datos II", "nota": 8.5, "edad": 19, "traslado_ugb": 15, "uso_telefono": 4.0},
}


sistema = SistemaEstadistico(dataset_ugb)

sistema.calcular_todo("nota")
sistema.calcular_todo("traslado_ugb")
sistema.calcular_todo("uso_telefono")
sistema.calcular_todo("edad")
```
