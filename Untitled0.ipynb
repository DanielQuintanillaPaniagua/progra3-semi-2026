```python
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
datos = pd.read_csv("/content/dataset.csv",sep=";",encoding="utf-8")
print(datos["celsius"])
```

```python
c = datos ["celsius"]
f = datos ["fahrenheit"]
print(c,f)
```

```python
modelo = tf.keras.Sequential()
modelo.add(tf.keras.layers.Dense(units=1,input_shape=[1]))
```

```python
modelo.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=1.0),loss="mean_squared_error")
```

```python
resp =modelo.predict(np.array([100.0]))
print(resp)
```
