```python
import tensorflow as tf
import pandas as pd
import numpy as np
```

```python
datos = pd.read_csv("/content/temperaturas.csv", sep=",", encoding="utf-8")
print(datos)
```

```python
c = datos["celsius"]
f = datos["fahrenheit"]
k = datos["kelvin"]
print(c, f, k)
```

```python
modelo_f = tf.keras.Sequential()
modelo_f.add(tf.keras.layers.Dense(units=1, input_shape=[1]))

modelo_k = tf.keras.Sequential()
modelo_k.add(tf.keras.layers.Dense(units=1, input_shape=[1]))
```

```python
modelo_f.compile(optimizer=tf.keras.optimizers.Adam(0.1), loss="mean_squared_error")
modelo_k.compile(optimizer=tf.keras.optimizers.Adam(0.1), loss="mean_squared_error")
```

```python
print("Entrenando Fahrenheit...")
historial_f = modelo_f.fit(c, f, epochs=1000, verbose=False)

print("Entrenando Kelvin...")
historial_k = modelo_k.fit(c, k, epochs=3000, verbose=False)

print("Listo")
```

```python
resp_f = modelo_f.predict(np.array([100.0]))
resp_k = modelo_k.predict(np.array([100.0]))
print("100 °C =", resp_f, "°F")
print("100 °C =", resp_k, "K")
```
